## 4. Data Cleaning

```python
#Since only four records are affected, replacing them with 0
df['children'] = df['children'].fillna(0) 

#Country is a categorical feature, so using the most frequent category (mode) is the standard approach.
mode = df["country"].mode()
df["country"] = df["country"].fillna(mode[0]) 

#An empty agent generally indicates that the booking was not made through a travel agent. Instead of deleting these rows, fill them with 0
df['agent'] = df['agent'].fillna(0) 

#Since almost the entire column is missing, it contributes very little useful information, so remove the column
df.drop(columns='company', inplace=True) 

#Removing Duplicates
df.drop_duplicates(inplace=True)

#Removing bookings where there are no guests
df = df[
    ~(
        (df["adults"] == 0) &
        (df["children"] == 0) &
        (df["babies"] == 0)
    )
].copy()

#Removing Negative Value Data
df = df[df["adr"] >= 0].copy()
```

### 5.1 Statistical Detection using IQR(Inter-Quartile Range)

```python
def find_outliers(column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1
    
    lower = Q1 - 1.5 * IQR
    upper = Q1 + 1.5 * IQR

    outliers = df[(df[column] < lower)|(df[column] > upper)]

    print(f"{column}")
    print("Lower Limit:", lower)
    print("Upper Limit:", upper)
    print("Number of Outliers:", len(outliers))
    print("-"*40)
```

```python
find_outliers('adr')
find_outliers('lead_time')
find_outliers('stays_in_week_nights')
find_outliers('adults')
```

### 5.2 Capping Extreme Values

```python
def cap_outliers(column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df[column] = df[column].clip(lower, upper)
```

```python
cap_outliers("adr")
cap_outliers("lead_time")
cap_outliers("stays_in_week_nights")
cap_outliers("stays_in_weekend_nights")
```

## 6. Data Type Conversion

```python
#Convert Categorical Columns
categorical_columns = [
    'hotel',
    'meal',
    'country',
    'market_segment',
    'distribution_channel',
    'reserved_room_type',
    'assigned_room_type',
    'deposit_type',
    'customer_type',
    'reservation_status'
]

for col in categorical_columns:
    df[col] = df[col].astype('category')

# Convert Float to Integer 
df["children"] = df["children"].astype(int)
df["agent"] = df["agent"].astype(int)

#Creating a single column using 3 different columns
df["arrival_date"] = pd.to_datetime(
    df["arrival_date_year"].astype(str) + "-" +
    df["arrival_date_month"] + "-" +
    df["arrival_date_day_of_month"].astype(str)
)
```

## 7. Feature Engineering

```python
#Instead of analyzing three separate columns, we can analysze by creating a single column feature.
df["total_guests"] = (
    df["adults"] +
    df["children"] +
    df["babies"]
)

#Create a feature representing the total stay duration
df["total_nights"] = (
    df["stays_in_week_nights"] +
    df["stays_in_weekend_nights"]
)

#Create a binary feature indicating whether the booking includes children or babies.
df["is_family"] = (
    (df["children"] > 0) |
    (df["babies"] > 0)
).astype(int)

#Convert the is_canceled column into a more readable format.
df["booking_status"] = df["is_canceled"].map({
    0: "Not Cancelled",
    1: "Cancelled"
})
```
